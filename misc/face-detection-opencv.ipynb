```python
image=cv2.imread("Image.jpg")
model=cv2.CascadeClassifier("model.xml")

face = model.detectMultiScale(image)
face2 = model.detectMultiScale(image)

x1=face[0][0]
y1=face[0][1]
a1=face[0][2]
b1=face[0][3]

x2=face[1][0]
y2=face[1][1]
a2=face[1][2]
b2=face[1][3]

image1 = cv2.rectangle(image , (x1,y1) , (x1+a1,y1+b1), (0,255,0),4)
image2 = cv2.rectangle(image , (x2,y2) , (x2+a2,y2+b2), (255,0,0),4)

image1 = cv2.cvtColor(image , cv2.COLOR_BGR2RGB)
image2 = cv2.cvtColor(image , cv2.COLOR_BGR2RGB)

plt.imshow(image1)
plt.imshow(image2)
```
