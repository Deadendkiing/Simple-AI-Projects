```python
import math
import pandas as pd

data = pd.read_csv('data/pima-indians-diabetes.data.csv')
data.head()
```

```python
# انتخاب سطر
record1 = data.iloc[0].values  # رکورد اول
record2 = data.iloc[1].values  # رکورد دوم
print(record1)
print(record2)
```

```python
# تابع برای محاسبه فاصله اقلیدوسی
def euclidean_distance(point1, point2):
    return math.sqrt(sum((x - y) ** 2 for x, y in zip(point1, point2)))
```

```python
# تابع برای محاسبه فاصله منهتن
def manhattan_distance(point1, point2):
    return sum(abs(x - y) for x, y in zip(point1, point2))
```

```python
# تابع برای محاسبه فاصله مینکوفسکی
def minkowski_distance(point1, point2, p):
    return sum(abs(x - y) ** p for x, y in zip(point1, point2)) ** (1 / p)
```

```python
# محاسبه فواصل
euclidean = euclidean_distance(record1, record2)
manhattan = manhattan_distance(record1, record2)
minkowski_p3 = minkowski_distance(record1, record2, p=3)
```

```python
# چاپ نتایج
print(f"رکورد اول: {record1}")
print(f"رکورد دوم: {record2}")
print(f"فاصله اقلیدوسی: {euclidean:.2f}")
print(f"فاصله منهتن: {manhattan:.2f}")
print(f"فاصله مینکوفسکی (p=3): {minkowski_p3:.2f}")
```
