```python
from tensorflow.keras.datasets import mnist
from sklearn.model_selection import train_test_split

# لود دیتاست MNIST
(X, y), _ = mnist.load_data()  # فقط داده‌های آموزشی اولیه رو لود می‌کنیم
X = X.reshape(-1, 28*28) / 255.0  # نرمال‌سازی

# تقسیم داده‌ها
X_temp, X_test, y_temp, y_test = train_test_split(X, y, test_size=0.1, random_state=42)
X_train, X_val, y_train, y_val = train_test_split(X_temp, y_temp, test_size=0.1111, random_state=42)  # 10% از 90% باقی‌مونده

# بررسی شکل داده‌ها
print("X_train shape:", X_train.shape)
print("X_val shape:", X_val.shape)
print("X_test shape:", X_test.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score


# 1. آموزش Random Forest
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)
rf_classifier.fit(X_train, y_train)
rf_val_accuracy = rf_classifier.score(X_val, y_val)
rf_test_accuracy = rf_classifier.score(X_test, y_test)
print("Random Forest Validation Accuracy:", rf_val_accuracy)
print("Random Forest Test Accuracy:", rf_test_accuracy)
```

```python
# 2. آموزش KNN
knn_classifier = KNeighborsClassifier(n_neighbors=3)
knn_classifier.fit(X_train, y_train)
knn_val_accuracy = knn_classifier.score(X_val, y_val)
knn_test_accuracy = knn_classifier.score(X_test, y_test)
print("KNN Validation Accuracy:", knn_val_accuracy)
print("KNN Test Accuracy:", knn_test_accuracy)
```

```python
# 3. آموزش SVM
svm_classifier = SVC(kernel='rbf', random_state=42)
svm_classifier.fit(X_train, y_train)
svm_val_accuracy = svm_classifier.score(X_val, y_val)
svm_test_accuracy = svm_classifier.score(X_test, y_test)
print("SVM Validation Accuracy:", svm_val_accuracy)
print("SVM Test Accuracy:", svm_test_accuracy)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score
```

```python
# 1. بهینه‌سازی Random Forest
rf_param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5]
}
rf_classifier = RandomForestClassifier(random_state=42)
rf_grid_search = GridSearchCV(rf_classifier, rf_param_grid, cv=3, scoring='accuracy')
rf_grid_search.fit(X_train, y_train)
print("Best Random Forest Params:", rf_grid_search.best_params_)
rf_best = rf_grid_search.best_estimator_
rf_val_accuracy = rf_best.score(X_val, y_val)
rf_test_accuracy = rf_best.score(X_test, y_test)
print("Optimized Random Forest Validation Accuracy:", rf_val_accuracy)
print("Optimized Random Forest Test Accuracy:", rf_test_accuracy)
```

```python
# 2. بهینه‌سازی KNN
knn_param_grid = {
    'n_neighbors': [3, 5, 7, 9],
    'weights': ['uniform', 'distance'],
    'p': [1, 2]  # 1 برای Manhattan، 2 برای Euclidean
}
knn_classifier = KNeighborsClassifier()
knn_grid_search = GridSearchCV(knn_classifier, knn_param_grid, cv=3, scoring='accuracy')
knn_grid_search.fit(X_train, y_train)
print("Best KNN Params:", knn_grid_search.best_params_)
knn_best = knn_grid_search.best_estimator_
knn_val_accuracy = knn_best.score(X_val, y_val)
knn_test_accuracy = knn_best.score(X_test, y_test)
print("Optimized KNN Validation Accuracy:", knn_val_accuracy)
print("Optimized KNN Test Accuracy:", knn_test_accuracy)
```

```python
# 3. بهینه‌سازی SVM
svm_param_grid = {
    'C': [0.1, 1, 10],
    'kernel': ['rbf', 'linear'],
    'gamma': ['scale', 'auto', 0.1]
}
svm_classifier = SVC(random_state=42)
svm_grid_search = GridSearchCV(svm_classifier, svm_param_grid, cv=3, scoring='accuracy')
svm_grid_search.fit(X_train, y_train)
print("Best SVM Params:", svm_grid_search.best_params_)
svm_best = svm_grid_search.best_estimator_
svm_val_accuracy = svm_best.score(X_val, y_val)
svm_test_accuracy = svm_best.score(X_test, y_test)
print("Optimized SVM Validation Accuracy:", svm_val_accuracy)
print("Optimized SVM Test Accuracy:", svm_test_accuracy)
```

```python
from sklearn.metrics import confusion_matrix
import numpy as np

# 1. ماتریس درهم‌ریختگی برای Random Forest
rf_predictions = rf_best.predict(X_test)
rf_conf_matrix = confusion_matrix(y_test, rf_predictions)
print("Random Forest Confusion Matrix:\n", rf_conf_matrix)
```

```python
# 2. ماتریس درهم‌ریختگی برای KNN
knn_predictions = knn_best.predict(X_test)
knn_conf_matrix = confusion_matrix(y_test, knn_predictions)
print("KNN Confusion Matrix:\n", knn_conf_matrix)
```

```python
# 3. ماتریس درهم‌ریختگی برای SVM
svm_predictions = svm_best.predict(X_test)
svm_conf_matrix = confusion_matrix(y_test, svm_predictions)
print("SVM Confusion Matrix:\n", svm_conf_matrix)
```

```python
from sklearn.ensemble import VotingClassifier
from sklearn.metrics import accuracy_score

# ایجاد Voting Classifier
voting_clf = VotingClassifier(
    estimators=[
        ('rf', rf_best),
        ('knn', knn_best),
        ('svm', svm_best)
    ],
    voting='hard'
)

# آموزش مدل ترکیبی روی داده‌های آموزشی
voting_clf.fit(X_train, y_train)

# ارزیابی روی مجموعه اعتبارسنجی
voting_val_accuracy = voting_clf.score(X_val, y_val)
print("Voting Classifier Validation Accuracy:", voting_val_accuracy)

# ارزیابی روی مجموعه تست (برای مقایسه نهایی)
voting_test_accuracy = voting_clf.score(X_test, y_test)
print("Voting Classifier Test Accuracy:", voting_test_accuracy)

# مقایسه با دقت مدل‌های جداگانه روی اعتبارسنجی
rf_val_accuracy = rf_best.score(X_val, y_val)
knn_val_accuracy = knn_best.score(X_val, y_val)
svm_val_accuracy = svm_best.score(X_val, y_val)
print("Random Forest Validation Accuracy:", rf_val_accuracy)
print("KNN Validation Accuracy:", knn_val_accuracy)
print("SVM Validation Accuracy:", svm_val_accuracy)
```

```python
from sklearn.metrics import accuracy_score

# 1. ارزیابی مدل ترکیبی روی مجموعه آزمایشی
voting_test_accuracy = voting_clf.score(X_test, y_test)
print("Voting Classifier Test Accuracy:", voting_test_accuracy)

# 2. ارزیابی مدل‌های جداگانه روی مجموعه آزمایشی
rf_test_accuracy = rf_best.score(X_test, y_test)
knn_test_accuracy = knn_best.score(X_test, y_test)
svm_test_accuracy = svm_best.score(X_test, y_test)

# 3. چاپ نتایج برای مقایسه
print("Random Forest Test Accuracy:", rf_test_accuracy)
print("KNN Test Accuracy:", knn_test_accuracy)
print("SVM Test Accuracy:", svm_test_accuracy)
print("Voting Classifier Test Accuracy:", voting_test_accuracy)
```

```python
import matplotlib.pyplot as plt

models = ['Random Forest', 'KNN', 'SVM', 'Voting']
accuracies = [rf_test_accuracy, knn_test_accuracy, svm_test_accuracy, voting_test_accuracy]

plt.bar(models, accuracies, color=['blue', 'green', 'red', 'purple'])
plt.title('Comparison of Model Accuracies on Test Set')
plt.ylabel('Accuracy')
plt.ylim(0, 1)  # محدوده دقت بین 0 و 1
for i, v in enumerate(accuracies):
    plt.text(i, v + 0.01, f'{v:.4f}', ha='center')
plt.show()
```
