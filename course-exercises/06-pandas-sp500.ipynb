```python
import pandas as pd
from pathlib import Path

%matplotlib inline
```

import matplotlib.pyplot as plt --->  not inline

%matplotlib inline ---> inline

```python
file_path = Path('data/sp500_companies.csv')

sp500_companies_csv = pd.read_csv(file_path)

sp500_companies_csv.head()

#Here we are adding our file's path and reading the contents using pandas 'read' function
```

```python
sector_count = sp500_companies_csv["Sector"].value_counts()
sector_count
#using this, we can see how many values are inside our file by column name
#for example here this code shows how many diffrent data we have based on their Sector
```

```python
sector_count.plot(kind='pie')
#using 'plot' , we can visualize our data in diffrent graph types
```

```python
#When we are working with big data, We need to filter our data to make it easier to work with
#in this example we are going to show 15 companies

market_cap =sp500_companies_csv.loc[:,["Symbol","Market Cap"]]
# first we extract the data we want

market_cap.set_index(market_cap["Symbol"], inplace=True)
# Second we set the index for our data

market_cap.drop(columns = ["Symbol"], inplace=True)
# we remove data that we dont want

top_15 = market_cap.nlargest(15, "Market Cap")
# extract top 15 by the numbers we want
top_15
top_15.plot(kind='bar' ,title = "Top 20 Companies in Market Cap in 2022")

top_15.plot(kind='scatter',x="Earnings/Share" , y="Price")
```
