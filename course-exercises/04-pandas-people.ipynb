```python
import pandas as pd
from pathlib import Path

%matplotlib inline
```

```python
file_path = Path("data/people.csv")

people = pd.read_csv(file_path)
people.head()
```

```python
people_d = people.drop(columns = ['web' , 'email' , 'phone1' , 'phone2' , 'address' , 'city'])
people_d.head()
```

```python
company_name = people_d['company_name'].value_counts()
company_name.head()
```

```python
county = people_d['county'].value_counts()
county.head()
```

```python
gender_count = people_d['gender'].value_counts()
gender_count.plot(kind='pie',title='Gender َAbundance')
```

```python
age_count = people_d['age'].value_counts()
age_count.describe()

# age_count.plot(kind = '') ????
```
