```python
import pandas as pd
from pathlib import Path

%matplotlib inline
```

```python
file_path = Path('data/customers.csv')

customers = pd.read_csv(file_path)

customers.head()
```

```python
balance = customers.loc[:,["Zip" , "Balance"]]

balance.set_index(customers["Zip"] , inplace = True)

balance.drop(columns = ["Zip"] , inplace = True)

print(balance.head())

top_20_balance = balance.nlargest(20 , "Balance")

# top_20_balance.plot(kind = "scatter" , x="Zip" , y="Balance") #Error???
```
