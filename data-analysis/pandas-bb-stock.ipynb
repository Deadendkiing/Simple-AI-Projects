```python
import pandas as pd
from pathlib import Path

%matplotlib inline
```

```python
file_path = Path ('data/bb_stock_data.csv')
```

```python
bb_df = pd.read_csv(file_path)
bb_df.head()
```

```python
bb_df = bb_df.set_index("Date")
bb_df.head()
```

```python
# @title Close

from matplotlib import pyplot as plt
bb_df['Close'].plot(kind='line', figsize=(8, 4), title='Close')
plt.gca().spines[['top', 'right']].set_visible(False)
```

```python
#Checking For Empty/Null Data

bb_df.isnull().mean() * 100
# Close Means There Are No Null Data
bb_df.isnull().sum()
#SUM Of Null Data Is 0 Means There Are No Null Data
```

```python
#Removing Empty/Null Data If There Are Any

bb_df = bb_df.dropna()
#After This We Can Check If Any Null Data Is Remaining
```

```python
#Returns a DataFrame with the percentage difference between the values for each row and, by default, the previous row
daily_returns =  bb_df.pct_change()
daily_returns.head()
```

```python
#Sorting Data By From Top
top = daily_returns.sort_values("Close" , ascending= False)
top.head()
```

```python
#Slicing Top 5 Records

top_5 = top.iloc[:5]

top_5.head()

top_5.plot()
```
