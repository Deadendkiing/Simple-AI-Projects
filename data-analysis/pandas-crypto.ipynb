```python
import pandas as pd
from pathlib import Path

%matplotlib inline
```

```python
file_path = Path("data/crypto_data.csv")
```

```python
#We Can Set An Index From Here And Parse Our Dates
crypto = pd.read_csv(file_path , index_col = "data_date" ,parse_dates = True)
crypto.head()
```

```python
crypto_nn = crypto.drop(columns = ["data_time" , "timestamp"])
crypto_nn.head()
```

```python
#Dropping Null Data
# crypto_nn = crypto.dropna()
#Its Better To Assign A New Variable
crypto_nn.isnull().sum()
```

```python
#Grouping By Currency And Plotting

crypto_plot = crypto_nn.groupby("cryptocurrency")["data_price"].plot(legend = True)
```

```python
#Two Years Average
crypto_av =crypto_nn.groupby("cryptocurrency")["data_price"].mean()
crypto_av.plot(kind = "bar")
```

```python
crypto_max = crypto_nn.groupby("cryptocurrency")['data_price'].max()
crypto_max.plot(kind = 'bar')
```

```python
crypto_min = crypto_nn.groupby("cryptocurrency")['data_price'].min()
crypto_min.plot(kind = 'bar')
```
